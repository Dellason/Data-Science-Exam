## 2  Clean & Prepare  <a id='clean'></a>

````python
# ── Missing values ──
df.isnull().sum()                          # missing count per column
df.isnull().sum() / len(df)               # missing % per column
df.dropna()                               # drop rows with ANY missing
df.dropna(subset=["col"])                 # drop only if 'col' is missing
df["col"].fillna(df["col"].mean())        # fill with mean
df["col"].fillna(df["col"].median())      # fill with median
df["col"].fillna(method="ffill")          # LOCF – carry last value forward
df["col"].fillna(method="bfill")          # NOCB – carry next value backward
df["col"].interpolate()                   # linear interpolation

# Fill NaN by group mean (e.g. fill Age by passenger class)
pclass_means = df.groupby("Pclass")["Age"].transform("mean")
df["Age"] = df["Age"].fillna(pclass_means)

# ── Drop columns / rows ──
df.drop(["col1", "col2"], axis=1)         # drop columns  (axis=1)
df.drop([0, 1, 2],        axis=0)         # drop rows     (axis=0)

# ── Filter rows ──
df[df["col"] > 100]
df[(df["A"] > 1) & (df["B"] == "X")]     # AND
df[(df["A"] > 1) | (df["B"] == "X")]     # OR
df[df["col"].isna()]                      # rows where col is NaN
df[df.isna().any(axis=1)]                 # rows with ANY missing value

# ── Rename / cast ──
df.rename(columns={"old": "new"}, inplace=True)
df["col"] = df["col"].astype(int)
````

## 5  Time Series  <a id='timeseries'></a>

````python
# ── Load ──
df = pd.read_csv("ts.csv", parse_dates=["date"], index_col="date")

# ── Slice by date ──
df.loc["2024"]                        # full year
df.loc["2024-03"]                     # one month
df.loc["2024-04-01":"2024-04-30"]     # date range

# ── Resample (change frequency) ──
df.resample("ME").mean()              # daily → monthly mean (ME = month end)
df.resample("YE").sum()               # → annual total
df.resample("W").mean()               # → weekly mean

# ── Group by year / decade ──
df.groupby(df.index.year).sum()

def to_decade(d): return (d.year // 10) * 10
df.groupby(to_decade).sum()

# ── Moving average (rolling mean) ──
df["col"].rolling(7).mean()           # 7-period rolling mean
df["col"].rolling(30).mean()          # 30-period rolling mean
# NOTE: larger window = smoother, but first (window-1) values become NaN

# ── Missing value imputation ──
df.fillna(method="ffill")             # LOCF: carry last value forward
df.fillna(method="bfill")             # NOCB: carry next value backward
df.interpolate()                      # linear interpolation

# ── Outlier detection: values outside (μ ± n·σ) are outliers ──
mu, sigma, n = df["col"].mean(), df["col"].std(), 2
outliers = df[(df["col"] < mu - n*sigma) | (df["col"] > mu + n*sigma)]

# ── datetime access ──
from datetime import datetime
d = datetime.now()
d.year, d.month, d.day, d.hour, d.minute   # access parts
d.date(); d.time()                          # date-only / time-only objects
````

## 9  Evaluation  <a id='evaluation'></a>

````python
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, ConfusionMatrixDisplay, classification_report
)

# ── Core metrics ──
# Accuracy:  fraction of correct predictions
# Precision: of predicted positives, how many are correct
# Recall:    of actual positives, how many were found
# F1:        harmonic mean of precision and recall

print("Accuracy: ",  accuracy_score(y_test, y_pred))
print("Precision:",  precision_score(y_test, y_pred, average='weighted', zero_division=0))
print("Recall:   ",  recall_score(y_test, y_pred, average='weighted', zero_division=0))
print("F1 Score: ",  f1_score(y_test, y_pred, average='weighted', zero_division=0))

# ── Evaluate function (reuse for multiple models) ──
def evaluate(y_true, y_pred, name):
    print(f"\n── {name} ──")
    print(f"Accuracy:  {accuracy_score(y_true, y_pred):.3f}")
    print(f"Precision: {precision_score(y_true, y_pred, average='weighted', zero_division=0):.3f}")
    print(f"Recall:    {recall_score(y_true, y_pred, average='weighted', zero_division=0):.3f}")
    print(f"F1 Score:  {f1_score(y_true, y_pred, average='weighted', zero_division=0):.3f}")

# ── Confusion matrix ──
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=model.classes_)
disp.plot(cmap="Blues"); plt.show()

# ── Full report ──
print(classification_report(y_test, y_pred, target_names=model.classes_))
````

## 10  NLP / Text  <a id='nlp'></a>

````python
import re
from bs4 import BeautifulSoup
from sklearn.feature_extraction.text import TfidfVectorizer

# ── Strip HTML tags ──
def html_to_text(html):
    if not isinstance(html, str): return ""
    return BeautifulSoup(html, "html.parser").get_text(separator=" ")

# ── Clean text ──
def clean_text(text):
    text = str(text).lower()              # lowercase
    text = re.sub(r"[^a-z\s]", "", text)  # keep only letters + spaces
    return text

# ── Remove special chars only (keep numbers) ──
def remove_special(text):
    if isinstance(text, str):
        return re.sub(r"[^a-zA-Z0-9\s]", "", text)
    return text

# ── Apply to DataFrame column ──
df["text"] = df["text"].apply(html_to_text)
df["text"] = df["text"].apply(clean_text)

# ── TF-IDF vectorisation ──
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

tfidf = TfidfVectorizer(max_features=5000, stop_words='english')
X_train = tfidf.fit_transform(train_df["text"])   # fit on train only
X_test  = tfidf.transform(test_df["text"])         # transform only (no refit!)
y_train = train_df["label"]
y_test  = test_df["label"]

# ── Train classifier on text features ──
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

log_model  = LogisticRegression(max_iter=1000)
log_model.fit(X_train, y_train)
y_pred_log = log_model.predict(X_test)

tree_model  = DecisionTreeClassifier(max_depth=10, random_state=42)
tree_model.fit(X_train, y_train)
y_pred_tree = tree_model.predict(X_test)

evaluate(y_test, y_pred_log,  "Logistic Regression")
evaluate(y_test, y_pred_tree, "Decision Tree")

# ── Predict on unseen data using SAME tfidf ──
X_unseen = tfidf.transform(unseen_df["text"])     # never refit tfidf!
unseen_df["predicted"] = log_model.predict(X_unseen)
````

## 11  Feature Selection & Importance  <a id='features'></a>

````python
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.inspection import permutation_importance
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=3)

# ── Forward selection: start empty, add best feature each step ──
sfs_fwd = SequentialFeatureSelector(
    knn, n_features_to_select=3, direction="forward"
).fit(X, y)
selected_fwd = list(X.columns[sfs_fwd.get_support()])
print("Forward selected:", selected_fwd)

# ── Backward selection: start with all, remove worst each step ──
sfs_bwd = SequentialFeatureSelector(
    knn, n_features_to_select=3, direction="backward"
).fit(X, y)
selected_bwd = list(X.columns[sfs_bwd.get_support()])
print("Backward selected:", selected_bwd)

# ── Permutation importance: how much accuracy drops when feature is shuffled ──
knn.fit(X, y)
result = permutation_importance(knn, X, y, n_repeats=10, random_state=0)
importances = result.importances_mean
sorted_idx  = importances.argsort()

plt.figure(figsize=(8, 5))
plt.barh(X.columns[sorted_idx], importances[sorted_idx], color="darkorange", zorder=3)
plt.xlabel("Permutation Importance Score")
plt.grid(axis="x"); plt.tight_layout(); plt.show()
````

## 12  Quick Reference

### f-string formatting
```python
f"{name:*^15}"      # centre in 15 chars padded with *
f"{n:>8.1f}"        # right-align, 8 wide, 1 decimal
f"{n:,.2f}"         # thousands separator, 2 decimals
f"{n:.2e}"          # scientific notation
f"{n:.1%}"          # percentage
```

### Resample frequency codes
| Code | Meaning |
|---|---|
| `D` | Calendar day |
| `W` | Weekly |
| `ME` | Month end |
| `MS` | Month start |
| `YE` | Year end |

### Key reminders
- `fit_transform` on **train only** → `transform` on test/unseen
- sklearn needs **2D arrays** — use `df[["col"]].values` not `df["col"].values`
- After `get_dummies`, use `.reindex(columns=X_train.columns, fill_value=0)` on unseen data
- `display(df)` keeps DataFrame formatting; `print(df)` does not
- `axis=1` → columns; `axis=0` → rows
- `zorder=3` on bars keeps them in front of grid lines
- `cross_val_score` uses the **full** scaled dataset, not train/test split
